#shallow copy and deep copy
Shallo copy : in a nested list when we copy the list it points to the address of original list so if we make any changes to the list it is reflected to original list.
deep copy : where as when we dp deep copy it creates seperate list with different address.

```python
import copy
```

```python
l1=[[1,2,3],[4,5,6]]
shallow_copy=copy.copy(l1)
print(shallow_copy)
```

```python
l1[1][0]=100
print(shallow_copy)
```

```python
shallow_copy[1][1]=200
print(l1)
```

```python
deep_copy=copy.deepcopy(l1)
print(deep_copy)
```

```python
deep_copy[0][1]=300
print(deep_copy)
print(l1)
```

```python
#what all copies we do in data structures like list, set and dictionaries are shallow copy 
lst=[1,2,3]
l1=lst.copy()
print(l1)
st={1,2,3}
s1=st.copy()
print(s1)
dt={'a':1,'b':2}
d1=dt.copy()
print(d1)
```

```python
import copy 
def double_numbers(numbers):
    result=copy.copy(numbers)
    for i in range(len(result)):
        result[i]=result[i]*2
    return result
original = [10,20,30]
doubled = double_numbers(original)
print('original:', original)
print('doubled:',doubled)
```

```python
#fibonnaci series:(iterative approach)
def fib_ser(n):
    ser=[0,1]
    for i in range(2,n):
        n_next=ser[i-1]+ser[i-2]
        ser.append(n_next)
    return ser[:n]
```

```python
fib_ser(10)
```

```python
#reccursive approach :
def fib_se_re(n):
    if n<=1:
        return n
    else:
        return (fib_se_re(n-1))+(fib_se_re(n-2))

    
```

```python
def li_re(n):
    return [fib_se_re(i) for i in range(n)]
```

```python
 li_re(10)
```
